```python
# -*- coding: utf-8 -*-
"""
Enhanced Geophone Analysis with Complete Output Information
Fixed STFT implementation and includes all visualization/output from previous version
"""

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score, f1_score, precision_recall_fscore_support
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from datetime import datetime
from scipy.signal import butter, lfilter, stft
import json
import time
from torch.optim.lr_scheduler import ReduceLROnPlateau
from tabulate import tabulate

# For Google Colab
try:
    from google.colab import drive
    IN_COLAB = True
except:
    IN_COLAB = False

# Use CUDA if available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

#######################
# Configuration
#######################

class GeophoneConfig:
    """Configuration for 3-class geophone analysis"""
    def __init__(self, **kwargs):
        # Data parameters
        self.window_size = 1000
        self.stride = 400  # 60% overlap
        self.batch_size = 32

        # Training parameters
        self.epochs = 20
        self.learning_rate = 0.001
        self.weight_decay = 0.01
        self.test_size = 0.2  # 80/20 split
        self.val_size = 0.2

        # Model parameters
        self.model_name = "geophone_3class"
        self.random_seed = 42
        self.patience = 5
        self.hidden_size = 64
        self.dropout_rate = 0.5

        # Signal processing - INCLUDE ALL FREQUENCIES 0-200
        self.lowcut = 0.1  # Changed from 5 to include low frequencies
        self.highcut = 200
        self.sampling_rate = 1000

        # Training settings
        self.clip_gradients = True
        self.max_grad_norm = 0.5
        self.use_scheduler = True
        self.model_type = "lstm"
        self.use_stft = False
        self.stft_nperseg = 128  # Reduced from 256 for more time frames
        self.stft_noverlap = 96   # Increased overlap for more time frames
        self.n_folds = 5
        self.use_cross_validation = True
        self.num_classes = 3

        # NO DATA AUGMENTATION - keep data poor
        self.add_noise = False
        self.noise_factor = 0.0
        self.time_shift = False
        self.max_shift = 0

        for key, value in kwargs.items():
            if hasattr(self, key):
                setattr(self, key, value)

#######################
# Model Definitions
#######################

class GeophoneLSTM3Class(nn.Module):
    def __init__(self, config, input_shape=None):
        super().__init__()

        self.config = config
        self.use_stft = config.use_stft

        if self.use_stft and input_shape is not None:
            self.input_size = input_shape[1]
            self.lstm_input_size = self.input_size
        else:
            self.input_size = 1
            self.lstm_input_size = self.input_size

        self.hidden_size = config.hidden_size
        self.dropout_rate = config.dropout_rate

        # LSTM layers
        self.lstm1 = nn.LSTM(
            input_size=self.lstm_input_size,
            hidden_size=self.hidden_size,
            num_layers=1,
            batch_first=True,
            dropout=0,
            bidirectional=True
        )

        self.lstm2 = nn.LSTM(
            input_size=self.hidden_size * 2,  # bidirectional
            hidden_size=self.hidden_size // 2,
            num_layers=1,
            batch_first=True,
            dropout=0
        )

        self.batch_norm = nn.BatchNorm1d(self.hidden_size // 2)
        self.fc1 = nn.Linear(self.hidden_size // 2, 32)
        self.dropout1 = nn.Dropout(self.dropout_rate)
        self.fc2 = nn.Linear(32, config.num_classes)

    def forward(self, x):
        if self.use_stft:
            x = x.permute(0, 2, 1)
        else:
            x = x.permute(0, 2, 1)

        lstm_out1, _ = self.lstm1(x)
        lstm_out2, (h_n, _) = self.lstm2(lstm_out1)

        h_n = h_n.squeeze(0)
        h_n = self.batch_norm(h_n)

        out = F.relu(self.fc1(h_n))
        out = self.dropout1(out)
        output = self.fc2(out)

        return output


class GeophoneGRU3Class(nn.Module):
    def __init__(self, config, input_shape=None):
        super().__init__()

        self.config = config
        self.use_stft = config.use_stft

        if self.use_stft and input_shape is not None:
            self.input_size = input_shape[1]
            self.gru_input_size = self.input_size
        else:
            self.input_size = 1
            self.gru_input_size = self.input_size

        self.hidden_size = config.hidden_size
        self.dropout_rate = config.dropout_rate

        # GRU layers
        self.gru1 = nn.GRU(
            input_size=self.gru_input_size,
            hidden_size=self.hidden_size,
            num_layers=1,
            batch_first=True,
            dropout=0,
            bidirectional=True
        )

        self.gru2 = nn.GRU(
            input_size=self.hidden_size * 2,
            hidden_size=self.hidden_size // 2,
            num_layers=1,
            batch_first=True,
            dropout=0
        )

        self.batch_norm = nn.BatchNorm1d(self.hidden_size // 2)
        self.fc1 = nn.Linear(self.hidden_size // 2, 32)
        self.dropout1 = nn.Dropout(self.dropout_rate)
        self.fc2 = nn.Linear(32, config.num_classes)

    def forward(self, x):
        if self.use_stft:
            x = x.permute(0, 2, 1)
        else:
            x = x.permute(0, 2, 1)

        gru_out1, _ = self.gru1(x)
        gru_out2, h_n = self.gru2(gru_out1)

        h_n = h_n.squeeze(0)
        h_n = self.batch_norm(h_n)

        out = F.relu(self.fc1(h_n))
        out = self.dropout1(out)
        output = self.fc2(out)

        return output


class GeophoneCNN3Class(nn.Module):
    def __init__(self, config, input_shape=None):
        super().__init__()

        self.config = config
        self.use_stft = config.use_stft

        if self.use_stft and input_shape is not None:
            self.input_channels = input_shape[1]
            # For STFT, we have very few time frames (7), so adjust kernel sizes
            self.kernel_sizes = [3, 3, 3]
            self.pool_size = 2  # Smaller pooling for STFT
        else:
            self.input_channels = 1
            self.kernel_sizes = [8, 5, 3]
            self.pool_size = 4  # Regular pooling for time domain

        # CNN layers with adaptive kernel sizes
        self.conv1 = nn.Conv1d(self.input_channels, 32, kernel_size=self.kernel_sizes[0], padding='same')
        self.bn1 = nn.BatchNorm1d(32)
        self.conv2 = nn.Conv1d(32, 64, kernel_size=self.kernel_sizes[1], padding='same')
        self.bn2 = nn.BatchNorm1d(64)
        self.conv3 = nn.Conv1d(64, 128, kernel_size=self.kernel_sizes[2], padding='same')
        self.bn3 = nn.BatchNorm1d(128)

        self.global_avg_pool = nn.AdaptiveAvgPool1d(1)
        self.fc1 = nn.Linear(128, 48)
        self.dropout1 = nn.Dropout(config.dropout_rate)
        self.fc2 = nn.Linear(48, 32)
        self.dropout2 = nn.Dropout(config.dropout_rate)
        self.fc3 = nn.Linear(32, config.num_classes)

    def forward(self, x):
        # First conv block
        x = F.relu(self.bn1(self.conv1(x)))
        if x.size(2) > self.pool_size:  # Only pool if we have enough length
            x = F.max_pool1d(x, kernel_size=self.pool_size)

        # Second conv block
        x = F.relu(self.bn2(self.conv2(x)))
        if x.size(2) > self.pool_size:  # Only pool if we have enough length
            x = F.max_pool1d(x, kernel_size=self.pool_size)

        # Third conv block (no pooling)
        x = F.relu(self.bn3(self.conv3(x)))

        # Global pooling and FC layers
        x = self.global_avg_pool(x)
        x = x.view(x.size(0), -1)

        x = F.relu(self.fc1(x))
        x = self.dropout1(x)
        x = F.relu(self.fc2(x))
        x = self.dropout2(x)
        output = self.fc3(x)

        return output


#######################
# Data Processing
#######################

def butter_bandpass_filter(data, lowcut, highcut, fs, order=4):
    nyq = 0.5 * fs
    low = lowcut / nyq
    high = highcut / nyq
    b, a = butter(order, [low, high], btype='band')
    y = lfilter(b, a, data)
    return y


def apply_stft_preprocessing(data, config):
    f, t, Zxx = stft(data, fs=config.sampling_rate,
                     nperseg=config.stft_nperseg,
                     noverlap=config.stft_noverlap)
    stft_magnitude = np.abs(Zxx)
    stft_magnitude = np.log1p(stft_magnitude)
    return stft_magnitude, f, t


def load_and_prepare_data_3class(file_path, label, config, max_minutes=None):
    """Load data without any augmentation"""
    print(f"Loading data from {file_path}...")

    try:
        df = pd.read_csv(file_path)

        if 'time_s' in df.columns and 'amplitude' in df.columns:
            time = df['time_s'].values
            data = df['amplitude'].values
        else:
            time = df.iloc[:, 0].values
            data = df.iloc[:, 1].values

        if max_minutes:
            max_samples = int(max_minutes * 60 * config.sampling_rate)
            if len(data) > max_samples:
                data = data[:max_samples]
                time = time[:max_samples]

        print(f"Loaded {len(data)} samples")

        # Apply bandpass filter
        data = butter_bandpass_filter(data, config.lowcut, config.highcut, config.sampling_rate)

        # Normalize
        data_min = np.min(data)
        data_max = np.max(data)
        if data_max > data_min:
            data_normalized = 2 * (data - data_min) / (data_max - data_min) - 1
        else:
            data_normalized = np.zeros_like(data)

        segments = []
        segment_times = []

        window_samples = config.window_size
        stride_samples = config.stride

        for i in range(0, len(data_normalized) - window_samples + 1, stride_samples):
            segment = data_normalized[i:i+window_samples].copy()
            if len(segment) == window_samples:
                # NO AUGMENTATION - keep data pure
                if config.use_stft:
                    stft_mag, _, _ = apply_stft_preprocessing(segment, config)
                    segments.append(stft_mag)
                else:
                    segments.append(segment)

                segment_times.append(time[i])

        labels = np.full(len(segments), label)
        print(f"Created {len(segments)} segments with label {label}")

        if len(segments) == 0:
            print("Warning: No segments created!")
            return torch.tensor([]), torch.tensor([]), []

        # Convert to tensors - FIX FOR STFT
        try:
            if config.use_stft:
                # STFT segments
                segments_array = np.array(segments, dtype=np.float32)
                # segments_array shape: (n_segments, n_freq, n_time)
                segments_tensor = torch.FloatTensor(segments_array)
            else:
                # Time domain segments
                segments_array = np.array(segments, dtype=np.float32)
                # Add channel dimension
                segments_tensor = torch.FloatTensor(segments_array).unsqueeze(1)

            labels_tensor = torch.LongTensor(labels)

            print(f"Final tensor shape: {segments_tensor.shape}")

            return segments_tensor, labels_tensor, segment_times[:len(labels)]

        except Exception as e:
            print(f"Error converting to tensors: {e}")
            import traceback
            traceback.print_exc()
            return torch.tensor([]), torch.tensor([]), []

    except Exception as e:
        print(f"Error loading data: {e}")
        import traceback
        traceback.print_exc()
        return torch.tensor([]), torch.tensor([]), []


class GeophoneDataset3Class(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


#######################
# Training Functions
#######################

def train_model_3class(model, train_loader, val_loader, config, model_path=None):
    """Train with detailed output like previous version"""
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.parameters(), lr=config.learning_rate, weight_decay=config.weight_decay)
    scheduler = ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=2, verbose=True)

    print("Starting training...")
    start_time = time.time()
    best_val_loss = float('inf')
    patience_counter = 0

    train_losses = []
    val_losses = []
    train_accuracies = []
    val_accuracies = []

    for epoch in range(config.epochs):
        model.train()
        epoch_loss = 0
        train_correct = 0
        train_total = 0

        for batch_idx, (inputs, targets) in enumerate(train_loader):
            inputs = inputs.to(device)
            targets = targets.to(device)

            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, targets)
            loss.backward()

            if config.clip_gradients:
                torch.nn.utils.clip_grad_norm_(model.parameters(), config.max_grad_norm)

            optimizer.step()
            epoch_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)
            train_total += targets.size(0)
            train_correct += (predicted == targets).sum().item()

            # Print batch progress like previous version
            if (batch_idx + 1) % 10 == 0:
                print(f"Epoch [{epoch+1}/{config.epochs}], Batch [{batch_idx+1}/{len(train_loader)}], Loss: {loss.item():.4f}")

        train_loss = epoch_loss / len(train_loader)
        train_accuracy = train_correct / train_total
        train_losses.append(train_loss)
        train_accuracies.append(train_accuracy)

        # Validation
        model.eval()
        val_loss = 0
        val_correct = 0
        val_total = 0

        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs = inputs.to(device)
                targets = targets.to(device)

                outputs = model(inputs)
                loss = criterion(outputs, targets)

                val_loss += loss.item()
                _, predicted = torch.max(outputs.data, 1)
                val_total += targets.size(0)
                val_correct += (predicted == targets).sum().item()

        val_loss = val_loss / len(val_loader)
        val_accuracy = val_correct / val_total
        val_losses.append(val_loss)
        val_accuracies.append(val_accuracy)

        scheduler.step(val_loss)

        print(f'Epoch [{epoch+1}/{config.epochs}], Train Loss: {train_loss:.4f}, Train Accuracy: {train_accuracy:.4f}, Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}')

        # Save best model
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            if model_path:
                torch.save({
                    'model_state_dict': model.state_dict(),
                    'optimizer_state_dict': optimizer.state_dict(),
                    'epoch': epoch,
                    'loss': val_loss,
                }, model_path)
                print(f"Model saved to {model_path}")
                print(f"Saved model checkpoint with validation loss: {val_loss:.4f}")
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= config.patience:
            print(f"Early stopping triggered after {epoch+1} epochs")
            break

    training_time = time.time() - start_time
    print(f"Training completed in {training_time:.2f} seconds")
    print(f"\nTraining complete. Best validation loss: {best_val_loss:.4f}")

    return model, train_losses, val_losses, train_accuracies, val_accuracies


#######################
# Visualization Functions
#######################

def plot_training_history(train_losses, val_losses, train_accuracies, val_accuracies, title):
    """Plot loss and accuracy over epochs"""
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

    # Loss plot
    ax1.plot(train_losses, label='Train Loss', color='blue')
    ax1.plot(val_losses, label='Val Loss', color='orange')
    ax1.set_xlabel('Epoch')
    ax1.set_ylabel('Loss')
    ax1.set_title(f'{title} - Loss over Epochs')
    ax1.legend()
    ax1.grid(True)

    # Accuracy plot
    ax2.plot(train_accuracies, label='Train Accuracy', color='green')
    ax2.plot(val_accuracies, label='Val Accuracy', color='red')
    ax2.set_xlabel('Epoch')
    ax2.set_ylabel('Accuracy')
    ax2.set_title(f'{title} - Accuracy over Epochs')
    ax2.legend()
    ax2.grid(True)

    plt.tight_layout()
    plt.show()


def create_classification_report_plot(precision, recall, f1, support, class_names, title, save_path=None):
    """Create a matplotlib plot of the classification report"""
    # Calculate averages
    accuracy = np.sum(precision * support) / np.sum(support)
    macro_precision = np.mean(precision)
    macro_recall = np.mean(recall)
    macro_f1 = np.mean(f1)
    weighted_precision = np.average(precision, weights=support)
    weighted_recall = np.average(recall, weights=support)
    weighted_f1 = np.average(f1, weights=support)

    # Create DataFrame
    data = {
        'Class': class_names + ['', 'Accuracy', 'Macro Avg', 'Weighted Avg'],
        'Precision': [f'{p:.2f}' for p in precision] + ['', '', f'{macro_precision:.2f}', f'{weighted_precision:.2f}'],
        'Recall': [f'{r:.2f}' for r in recall] + ['', '', f'{macro_recall:.2f}', f'{weighted_recall:.2f}'],
        'F1-Score': [f'{f:.2f}' for f in f1] + ['', f'{accuracy:.2f}', f'{macro_f1:.2f}', f'{weighted_f1:.2f}'],
        'Support': [f'{int(s)}' for s in support] + ['', f'{int(np.sum(support))}', f'{int(np.sum(support))}', f'{int(np.sum(support))}']
    }
    df = pd.DataFrame(data)

    # Create figure
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.axis('off')

    # Table parameters
    table_width = 0.9
    table_height = 0.6
    x_start = 0.05
    y_start = 0.1

    # Title
    plt.figtext(0.5, 0.88, title, fontsize=10, ha='center', fontweight='bold')

    # Draw horizontal lines
    def add_hline(y_pos, linewidth=1):
        line = plt.Line2D([x_start, x_start + table_width],
                         [y_pos, y_pos],
                         transform=fig.transFigure,
                         color='black',
                         linewidth=linewidth)
        fig.add_artist(line)

    # Add lines
    add_hline(y_start + table_height, linewidth=1)
    add_hline(y_start, linewidth=1)

    # Add separator lines
    header_height = 0.08
    class_rows = len(class_names)
    add_hline(y_start + table_height - header_height, linewidth=1)
    add_hline(y_start + table_height - header_height - (class_rows * (table_height - header_height) / len(df)), linewidth=1)

    # Headers
    col_names = ['Class', 'Precision', 'Recall', 'F1-Score', 'Support']
    col_widths = [0.25, 0.16, 0.16, 0.16, 0.16]
    col_positions = [x_start]
    for w in col_widths[:-1]:
        col_positions.append(col_positions[-1] + w)

    for i, col in enumerate(col_names):
        plt.figtext(col_positions[i] + col_widths[i]/2,
                   y_start + table_height - header_height/2,
                   col,
                   fontweight='bold',
                   ha='center',
                   va='center')

    # Data rows
    row_height = (table_height - header_height) / len(df)
    for row_idx in range(len(df)):
        y_pos = y_start + table_height - header_height - (row_idx + 0.5) * row_height

        # Class name
        weight = 'bold' if row_idx >= class_rows + 1 else 'normal'
        plt.figtext(col_positions[0] + 0.01, y_pos, df.iloc[row_idx, 0],
                   fontweight=weight, ha='left', va='center')

        # Other columns
        for col_idx in range(1, len(col_names)):
            plt.figtext(col_positions[col_idx] + col_widths[col_idx]/2, y_pos,
                       df.iloc[row_idx, col_idx], ha='center', va='center')

    plt.tight_layout(rect=[0, 0, 1, 0.85])

    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()


def evaluate_model_3class(model, test_loader, class_names, plot_title="Classification Report"):
    """Evaluate model and create visualized report"""
    model.eval()
    all_preds = []
    all_targets = []
    test_loss = 0
    criterion = nn.CrossEntropyLoss()

    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs = inputs.to(device)
            targets = targets.to(device)

            outputs = model(inputs)
            loss = criterion(outputs, targets)
            test_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)

            all_preds.extend(predicted.cpu().numpy())
            all_targets.extend(targets.cpu().numpy())

    test_loss = test_loss / len(test_loader)

    # Calculate metrics
    accuracy = accuracy_score(all_targets, all_preds)
    precision, recall, f1, support = precision_recall_fscore_support(all_targets, all_preds, average=None)

    print(f"\nTest Loss: {test_loss:.4f}, Test Accuracy: {accuracy:.4f}")

    # Print classification report
    print("\nClassification Report:")
    print(classification_report(all_targets, all_preds, target_names=class_names))

    # Confusion matrix
    cm = confusion_matrix(all_targets, all_preds)

    # Create visualized classification report
    create_classification_report_plot(precision, recall, f1, support, class_names, plot_title)

    return {
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'support': support,
        'confusion_matrix': cm,
        'predictions': all_preds,
        'targets': all_targets,
        'test_loss': test_loss
    }


def plot_confusion_matrix_3class(cm, class_names, title):
    """Plot confusion matrix with better visualization"""
    plt.figure(figsize=(8, 6))

    # Normalize confusion matrix for color scaling
    cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    # Create heatmap
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names,
                yticklabels=class_names,
                cbar_kws={'label': 'Count'})

    # Add percentage annotations
    for i in range(len(class_names)):
        for j in range(len(class_names)):
            percentage = cm_normalized[i, j] * 100
            plt.text(j + 0.5, i + 0.7, f'({percentage:.1f}%)',
                    ha='center', va='center', fontsize=8, color='gray')

    plt.title(title)
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.tight_layout()
    plt.show()


def perform_5fold_cv_3class(X, y, config, model_class, class_names):
    """5-fold cross-validation with visualization"""
    kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=config.random_seed)

    fold_results = []
    all_fold_preds = []
    all_fold_targets = []

    print(f"\nPerforming 5-fold cross-validation on entire dataset ({len(X)} samples)...")

    for fold, (train_idx, val_idx) in enumerate(kf.split(X, y)):
        print(f"\n--- Fold {fold + 1}/5 ---")

        X_train_fold = X[train_idx]
        X_val_fold = X[val_idx]
        y_train_fold = y[train_idx]
        y_val_fold = y[val_idx]

        train_dataset = GeophoneDataset3Class(X_train_fold, y_train_fold)
        val_dataset = GeophoneDataset3Class(X_val_fold, y_val_fold)

        train_loader = DataLoader(train_dataset, batch_size=config.batch_size, shuffle=True)
        val_loader = DataLoader(val_dataset, batch_size=config.batch_size)

        # Create model
        input_shape = X.shape if config.use_stft else None
        model = model_class(config, input_shape).to(device)

        # Train
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(model.parameters(), lr=config.learning_rate, weight_decay=config.weight_decay)

        # Train for fewer epochs in CV
        for epoch in range(10):
            model.train()
            for inputs, targets in train_loader:
                inputs = inputs.to(device)
                targets = targets.to(device)

                optimizer.zero_grad()
                outputs = model(inputs)
                loss = criterion(outputs, targets)
                loss.backward()

                if config.clip_gradients:
                    torch.nn.utils.clip_grad_norm_(model.parameters(), config.max_grad_norm)

                optimizer.step()

        # Evaluate
        model.eval()
        val_preds = []
        val_targets = []

        with torch.no_grad():
            for inputs, targets in val_loader:
                inputs = inputs.to(device)
                targets = targets.to(device)

                outputs = model(inputs)
                _, predicted = torch.max(outputs.data, 1)

                val_preds.extend(predicted.cpu().numpy())
                val_targets.extend(targets.cpu().numpy())

        accuracy = accuracy_score(val_targets, val_preds)
        f1 = f1_score(val_targets, val_preds, average='weighted')

        fold_results.append({
            'fold': fold + 1,
            'accuracy': accuracy,
            'f1_score': f1,
            'predictions': val_preds,
            'targets': val_targets
        })

        all_fold_preds.extend(val_preds)
        all_fold_targets.extend(val_targets)

        print(f"Fold {fold + 1} - Accuracy: {accuracy:.4f}, F1-Score: {f1:.4f}")

    # Overall metrics
    overall_accuracy = accuracy_score(all_fold_targets, all_fold_preds)
    overall_f1 = f1_score(all_fold_targets, all_fold_preds, average='weighted')
    overall_cm = confusion_matrix(all_fold_targets, all_fold_preds)

    return {
        'fold_results': fold_results,
        'overall_accuracy': overall_accuracy,
        'overall_f1_score': overall_f1,
        'overall_confusion_matrix': overall_cm,
        'accuracy_mean': np.mean([f['accuracy'] for f in fold_results]),
        'accuracy_std': np.std([f['accuracy'] for f in fold_results]),
        'f1_mean': np.mean([f['f1_score'] for f in fold_results]),
        'f1_std': np.std([f['f1_score'] for f in fold_results])
    }


def plot_model_comparison(all_results):
    """Plot model accuracy comparison across different architectures"""
    # Extract data for plotting
    models = []
    scenarios = []
    accuracies = []

    for exp_name, results in all_results.items():
        if 'test_results' not in results:
            continue

        parts = exp_name.split('_')
        model = parts[0].upper()
        preprocessing = parts[1]
        scenario = ' '.join(parts[2:]).title()

        models.append(f"{model}\n{preprocessing}")
        scenarios.append(scenario)
        accuracies.append(results['test_results']['accuracy'])

    if not models:
        print("No complete results to plot.")
        return

    # Create grouped bar chart
    fig, ax = plt.subplots(figsize=(12, 6))

    # Group by scenario
    unique_scenarios = list(sorted(set(scenarios)))
    unique_models = list(sorted(set(models)))

    x = np.arange(len(unique_scenarios))
    width = 0.8 / len(unique_models)

    for i, model in enumerate(unique_models):
        model_accs = []
        for scenario in unique_scenarios:
            # Find accuracy for this model and scenario
            acc = 0
            for j, (m, s, a) in enumerate(zip(models, scenarios, accuracies)):
                if m == model and s == scenario:
                    acc = a
                    break
            model_accs.append(acc)

        ax.bar(x + i * width - width * len(unique_models) / 2, model_accs,
               width, label=model)

    ax.set_xlabel('Training Scenario')
    ax.set_ylabel('Test Accuracy')
    ax.set_title('Model Accuracy Comparison Across Different Architectures')
    ax.set_xticks(x)
    ax.set_xticklabels(unique_scenarios)
    ax.legend()
    ax.grid(True, axis='y', alpha=0.3)

    plt.tight_layout()
    plt.show()


#######################
# Main Experiment Runner
#######################

def run_all_experiments(drive_path):
    """Run all experiments with complete output information"""

    # Model configurations
    model_types = ['lstm', 'gru', 'cnn']
    use_stft_options = [False, True]
    training_scenarios = ['human_only', 'car_only', 'combined']

    # Class names
    class_names = ['Empty', 'Stomping', 'Car']

    all_results = {}

    for model_type in model_types:
        for use_stft in use_stft_options:
            for scenario in training_scenarios:

                preprocessing = "STFT" if use_stft else "TimeDomain"
                experiment_name = f"{model_type}_{preprocessing}_{scenario}"

                print(f"\n{'='*80}")
                print(f"EXPERIMENT: {experiment_name}")
                print(f"{'='*80}")

                # Create config
                config = GeophoneConfig(
                    model_type=model_type,
                    use_stft=use_stft,
                    epochs=20,
                    batch_size=32,
                    learning_rate=0.001,
                    dropout_rate=0.5,
                    hidden_size=64,
                    weight_decay=0.01,
                    add_noise=False,  # NO augmentation
                    lowcut=0.1,  # Include all frequencies from near 0
                    highcut=200
                )

                # Load data based on scenario
                if scenario == 'human_only':
                    human_segments, human_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'human.csv'), label=1, config=config)
                    nothing_segments, nothing_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'human_nothing.csv'), label=0, config=config)

                    if len(human_segments) == 0 or len(nothing_segments) == 0:
                        print(f"ERROR: No data loaded for {experiment_name}. Skipping...")
                        continue

                    X = torch.cat([nothing_segments, human_segments])
                    y = torch.cat([nothing_labels, human_labels])

                    config.num_classes = 2
                    current_class_names = ['Empty', 'Stomping']

                elif scenario == 'car_only':
                    car_segments, car_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'car.csv'), label=2, config=config)
                    nothing_segments, nothing_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'car_nothing.csv'), label=0, config=config)

                    if len(car_segments) == 0 or len(nothing_segments) == 0:
                        print(f"ERROR: No data loaded for {experiment_name}. Skipping...")
                        continue

                    X = torch.cat([nothing_segments, car_segments])
                    y = torch.cat([nothing_labels, car_labels])

                    # Remap labels for binary classification
                    y[y == 2] = 1
                    config.num_classes = 2
                    current_class_names = ['Empty', 'Car']

                else:  # combined
                    human_segments, human_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'human.csv'), label=1, config=config)
                    human_nothing_segments, human_nothing_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'human_nothing.csv'), label=0, config=config)
                    car_segments, car_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'car.csv'), label=2, config=config)
                    car_nothing_segments, car_nothing_labels, _ = load_and_prepare_data_3class(
                        os.path.join(drive_path, 'car_nothing.csv'), label=0, config=config)

                    if (len(human_segments) == 0 or len(car_segments) == 0 or
                        len(human_nothing_segments) == 0 or len(car_nothing_segments) == 0):
                        print(f"ERROR: No data loaded for {experiment_name}. Skipping...")
                        continue

                    X = torch.cat([human_nothing_segments, car_nothing_segments,
                                  human_segments, car_segments])
                    y = torch.cat([human_nothing_labels, car_nothing_labels,
                                  human_labels, car_labels])

                    config.num_classes = 3
                    current_class_names = class_names

                print(f"Dataset size: {len(X)} samples")
                print(f"Class distribution: {[(i, (y==i).sum().item()) for i in range(config.num_classes)]}")

                # Split data
                X_train, X_test, y_train, y_test = train_test_split(
                    X, y, test_size=config.test_size, random_state=42, stratify=y)

                X_train, X_val, y_train, y_val = train_test_split(
                    X_train, y_train, test_size=config.val_size, random_state=42, stratify=y_train)

                print(f"Data split - Train: {len(X_train)} ({len(X_train)/len(X)*100:.1f}%), Val: {len(X_val)} ({len(X_val)/len(X)*100:.1f}%), Test: {len(X_test)} ({len(X_test)/len(X)*100:.1f}%)")

                # Create datasets
                train_dataset = GeophoneDataset3Class(X_train, y_train)
                val_dataset = GeophoneDataset3Class(X_val, y_val)
                test_dataset = GeophoneDataset3Class(X_test, y_test)

                train_loader = DataLoader(train_dataset, batch_size=config.batch_size, shuffle=True)
                val_loader = DataLoader(val_dataset, batch_size=config.batch_size)
                test_loader = DataLoader(test_dataset, batch_size=config.batch_size)

                # Select model
                if model_type == 'lstm':
                    model_class = GeophoneLSTM3Class
                elif model_type == 'gru':
                    model_class = GeophoneGRU3Class
                else:
                    model_class = GeophoneCNN3Class

                # Create and train model
                input_shape = X_train.shape if use_stft else None
                model = model_class(config, input_shape).to(device)

                # Print model architecture
                print(f"\n{scenario.replace('_', ' ').title()} {model_type.upper()} Model Architecture:")
                print(model)

                model_path = os.path.join(drive_path, 'models', f'{experiment_name}.pth')
                os.makedirs(os.path.dirname(model_path), exist_ok=True)

                model, train_losses, val_losses, train_accs, val_accs = train_model_3class(
                    model, train_loader, val_loader, config, model_path)

                # Plot training history
                plot_training_history(train_losses, val_losses, train_accs, val_accs,
                                    f"{model_type.upper()} {preprocessing} - {scenario.replace('_', ' ').title()}")

                # Load best model
                checkpoint = torch.load(model_path, map_location=device)
                model.load_state_dict(checkpoint['model_state_dict'])

                # Evaluate
                print(f"\n=== Test Set Evaluation ===")
                plot_title = f"TABLE IV\nCLASSIFICATION REPORT FOR {model_type.upper()} MODEL ON TEST SET"
                test_results = evaluate_model_3class(model, test_loader, current_class_names, plot_title)

                # Plot confusion matrix
                plot_confusion_matrix_3class(
                    test_results['confusion_matrix'],
                    current_class_names,
                    f'Confusion Matrix - {experiment_name}'
                )

                # Perform 5-fold CV
                print(f"\n=== 5-Fold Cross-Validation ===")
                cv_results = perform_5fold_cv_3class(X, y, config, model_class, current_class_names)

                print(f"\nCV Results (on full dataset of {len(X)} samples):")
                print(f"Accuracy: {cv_results['accuracy_mean']:.4f} ± {cv_results['accuracy_std']:.4f}")
                print(f"F1-Score: {cv_results['f1_mean']:.4f} ± {cv_results['f1_std']:.4f}")

                # Store results
                all_results[experiment_name] = {
                    'test_results': test_results,
                    'cv_results': cv_results,
                    'config': config,
                    'train_losses': train_losses,
                    'val_losses': val_losses,
                    'train_accuracies': train_accs,
                    'val_accuracies': val_accs
                }

    return all_results


def create_summary_table(all_results):
    """Create summary table visualization"""
    print("\n" + "="*100)
    print("SUMMARY TABLE - ALL EXPERIMENTS")
    print("="*100)

    summary_data = []

    for exp_name, results in sorted(all_results.items()):
        if 'test_results' not in results:
            continue

        parts = exp_name.split('_')
        model = parts[0].upper()
        preprocessing = parts[1]
        scenario = ' '.join(parts[2:]).title()

        test_acc = results['test_results']['accuracy']
        test_f1 = np.mean(results['test_results']['f1'])

        # Handle cases where CV might not have completed
        if 'cv_results' in results:
            cv_acc = results['cv_results']['accuracy_mean']
            cv_acc_std = results['cv_results']['accuracy_std']
            cv_f1 = results['cv_results']['f1_mean']
            cv_f1_std = results['cv_results']['f1_std']
            cv_acc_str = f"{cv_acc:.3f} ± {cv_acc_std:.3f}"
            cv_f1_str = f"{cv_f1:.3f} ± {cv_f1_std:.3f}"
        else:
            cv_acc_str = "N/A"
            cv_f1_str = "N/A"

        summary_data.append([
            model, preprocessing, scenario,
            f"{test_acc:.3f}", f"{test_f1:.3f}",
            cv_acc_str, cv_f1_str
        ])

    if not summary_data:
        print("No complete results to display in summary table.")
        return

    # Create DataFrame for plotting
    df = pd.DataFrame(summary_data, columns=[
        "Model", "Preprocessing", "Training Scenario",
        "Test Acc", "Test F1", "CV Acc (Mean±Std)", "CV F1 (Mean±Std)"
    ])

    # Create figure
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.axis('off')

    # Create table
    table = ax.table(cellText=df.values,
                     colLabels=df.columns,
                     cellLoc='center',
                     loc='center',
                     bbox=[0, 0, 1, 1])

    # Style the table
    table.auto_set_font_size(False)
    table.set_fontsize(9)
    table.scale(1.2, 1.5)

    # Header styling
    for i in range(len(df.columns)):
        table[(0, i)].set_facecolor('#40466e')
        table[(0, i)].set_text_props(weight='bold', color='white')

    # Alternate row colors
    for i in range(1, len(df) + 1):
        if i % 2 == 0:
            for j in range(len(df.columns)):
                table[(i, j)].set_facecolor('#f1f1f2')

    plt.title('Experimental Results Summary', fontsize=16, fontweight='bold', pad=20)
    plt.tight_layout()
    plt.show()


def perform_cross_domain_evaluation(drive_path, all_results):
    """Perform cross-domain evaluation - test models on different data types"""
    print("\n" + "="*100)
    print("CROSS-DOMAIN EVALUATION")
    print("="*100)

    # Load all data for cross-domain testing
    config = GeophoneConfig(
        lowcut=0.1,
        highcut=200,
        add_noise=False,
        use_stft=False  # Use time domain for cross-domain testing
    )

    # Load human test data
    human_segments, _, _ = load_and_prepare_data_3class(
        os.path.join(drive_path, 'human.csv'), label=1, config=config)
    human_nothing_segments, _, _ = load_and_prepare_data_3class(
        os.path.join(drive_path, 'human_nothing.csv'), label=0, config=config)

    if len(human_segments) > 0 and len(human_nothing_segments) > 0:
        X_human = torch.cat([human_nothing_segments, human_segments])
        y_human = torch.cat([torch.zeros(len(human_nothing_segments), dtype=torch.long),
                            torch.ones(len(human_segments), dtype=torch.long)])

        _, X_human_test, _, y_human_test = train_test_split(
            X_human, y_human, test_size=0.2, random_state=42, stratify=y_human)

        human_test_dataset = GeophoneDataset3Class(X_human_test, y_human_test)
        human_test_loader = DataLoader(human_test_dataset, batch_size=32)

    # Load car test data
    car_segments, _, _ = load_and_prepare_data_3class(
        os.path.join(drive_path, 'car.csv'), label=2, config=config)
    car_nothing_segments, _, _ = load_and_prepare_data_3class(
        os.path.join(drive_path, 'car_nothing.csv'), label=0, config=config)

    if len(car_segments) > 0 and len(car_nothing_segments) > 0:
        X_car = torch.cat([car_nothing_segments, car_segments])
        y_car = torch.cat([torch.zeros(len(car_nothing_segments), dtype=torch.long),
                          torch.ones(len(car_segments), dtype=torch.long)])  # Map to 1 for binary

        _, X_car_test, _, y_car_test = train_test_split(
            X_car, y_car, test_size=0.2, random_state=42, stratify=y_car)

        car_test_dataset = GeophoneDataset3Class(X_car_test, y_car_test)
        car_test_loader = DataLoader(car_test_dataset, batch_size=32)

    # Test human model on car data
    if 'lstm_TimeDomain_human_only' in all_results:
        print("\n=== Evaluating Human LSTM Model on Car Test Data ===")
        model_path = os.path.join(drive_path, 'models', 'lstm_TimeDomain_human_only.pth')
        if os.path.exists(model_path):
            config.num_classes = 2
            model = GeophoneLSTM3Class(config).to(device)
            checkpoint = torch.load(model_path, map_location=device)
            model.load_state_dict(checkpoint['model_state_dict'])

            results = evaluate_model_3class(
                model, car_test_loader, ['Empty', 'Car'],
                "Human LSTM Model on Car Data"
            )

            print(f"Human LSTM Model on Car Data Loss: {results['test_loss']:.4f}, Accuracy: {results['accuracy']:.4f}")

    # Test car model on human data
    if 'lstm_TimeDomain_car_only' in all_results:
        print("\n=== Evaluating Car LSTM Model on Human Test Data ===")
        model_path = os.path.join(drive_path, 'models', 'lstm_TimeDomain_car_only.pth')
        if os.path.exists(model_path):
            config.num_classes = 2
            model = GeophoneLSTM3Class(config).to(device)
            checkpoint = torch.load(model_path, map_location=device)
            model.load_state_dict(checkpoint['model_state_dict'])

            results = evaluate_model_3class(
                model, human_test_loader, ['Empty', 'Stomping'],
                "Car LSTM Model on Human Data"
            )

            print(f"Car LSTM Model on Human Data Loss: {results['test_loss']:.4f}, Accuracy: {results['accuracy']:.4f}")

    # Test combined model on both
    if 'lstm_TimeDomain_combined' in all_results:
        print("\n=== Evaluating Combined LSTM Model on Human Test Data ===")
        model_path = os.path.join(drive_path, 'models', 'lstm_TimeDomain_combined.pth')
        if os.path.exists(model_path):
            config.num_classes = 3
            model = GeophoneLSTM3Class(config).to(device)
            checkpoint = torch.load(model_path, map_location=device)
            model.load_state_dict(checkpoint['model_state_dict'])

            # Special evaluation for 3-class model on 2-class data
            model.eval()
            all_preds = []
            all_targets = []
            test_loss = 0
            criterion = nn.CrossEntropyLoss()

            with torch.no_grad():
                for inputs, targets in human_test_loader:
                    inputs = inputs.to(device)
                    targets = targets.to(device)

                    outputs = model(inputs)
                    loss = criterion(outputs, targets)
                    test_loss += loss.item()

                    _, predicted = torch.max(outputs.data, 1)

                    all_preds.extend(predicted.cpu().numpy())
                    all_targets.extend(targets.cpu().numpy())

            test_loss = test_loss / len(human_test_loader)

            # Only evaluate on classes that exist in the test data
            unique_targets = np.unique(all_targets)
            accuracy = accuracy_score(all_targets, all_preds)

            print(f"Combined LSTM Model on Human Data Loss: {test_loss:.4f}, Accuracy: {accuracy:.4f}")
            print("\nClassification Report:")
            print(classification_report(all_targets, all_preds, labels=unique_targets,
                                      target_names=['Empty', 'Stomping']))

            print("\n=== Evaluating Combined LSTM Model on Car Test Data ===")
            # For car data, we need to remap labels (0 stays 0, 1 becomes 2)
            X_car_combined = X_car_test
            y_car_combined = y_car_test.clone()
            y_car_combined[y_car_combined == 1] = 2  # Map car to label 2

            car_combined_dataset = GeophoneDataset3Class(X_car_combined, y_car_combined)
            car_combined_loader = DataLoader(car_combined_dataset, batch_size=32)

            # Special evaluation for 3-class model on car data (classes 0 and 2)
            model.eval()
            all_preds = []
            all_targets = []
            test_loss = 0

            with torch.no_grad():
                for inputs, targets in car_combined_loader:
                    inputs = inputs.to(device)
                    targets = targets.to(device)

                    outputs = model(inputs)
                    loss = criterion(outputs, targets)
                    test_loss += loss.item()

                    _, predicted = torch.max(outputs.data, 1)

                    all_preds.extend(predicted.cpu().numpy())
                    all_targets.extend(targets.cpu().numpy())

            test_loss = test_loss / len(car_combined_loader)

            unique_targets = np.unique(all_targets)
            accuracy = accuracy_score(all_targets, all_preds)

            print(f"Combined LSTM Model on Car Data Loss: {test_loss:.4f}, Accuracy: {accuracy:.4f}")
            print("\nClassification Report:")
            # Map labels for display
            target_names = []
            for label in unique_targets:
                if label == 0:
                    target_names.append('Empty')
                elif label == 2:
                    target_names.append('Car')
            print(classification_report(all_targets, all_preds, labels=unique_targets,
                                      target_names=target_names))

    # Repeat for GRU models
    for model_type in ['gru', 'cnn']:
        if f'{model_type}_TimeDomain_human_only' in all_results:
            print(f"\n=== Cross-Domain {model_type.upper()} EVALUATION ===")

            # Get appropriate model class
            if model_type == 'gru':
                model_class = GeophoneGRU3Class
            else:
                model_class = GeophoneCNN3Class

            # Test human model on car data
            print(f"\n=== Evaluating Human {model_type.upper()} Model on Car Test Data ===")
            model_path = os.path.join(drive_path, 'models', f'{model_type}_TimeDomain_human_only.pth')
            if os.path.exists(model_path):
                config.num_classes = 2
                model = model_class(config).to(device)
                checkpoint = torch.load(model_path, map_location=device)
                model.load_state_dict(checkpoint['model_state_dict'])

                results = evaluate_model_3class(
                    model, car_test_loader, ['Empty', 'Car'],
                    f"Human {model_type.upper()} Model on Car Data"
                )

                print(f"Human {model_type.upper()} Model on Car Data Loss: {results['test_loss']:.4f}, Accuracy: {results['accuracy']:.4f}")

            # Test car model on human data
            print(f"\n=== Evaluating Car {model_type.upper()} Model on Human Test Data ===")
            model_path = os.path.join(drive_path, 'models', f'{model_type}_TimeDomain_car_only.pth')
            if os.path.exists(model_path):
                config.num_classes = 2
                model = model_class(config).to(device)
                checkpoint = torch.load(model_path, map_location=device)
                model.load_state_dict(checkpoint['model_state_dict'])

                results = evaluate_model_3class(
                    model, human_test_loader, ['Empty', 'Stomping'],
                    f"Car {model_type.upper()} Model on Human Data"
                )

                print(f"Car {model_type.upper()} Model on Human Data Loss: {results['test_loss']:.4f}, Accuracy: {results['accuracy']:.4f}")

            # Test combined models
            print(f"\n=== Evaluating Combined {model_type.upper()} Model on Human Test Data ===")
            model_path = os.path.join(drive_path, 'models', f'{model_type}_TimeDomain_combined.pth')
            if os.path.exists(model_path):
                config.num_classes = 3
                model = model_class(config).to(device)
                checkpoint = torch.load(model_path, map_location=device)
                model.load_state_dict(checkpoint['model_state_dict'])

                # Special evaluation for 3-class model on 2-class data
                model.eval()
                all_preds = []
                all_targets = []
                test_loss = 0
                criterion = nn.CrossEntropyLoss()

                with torch.no_grad():
                    for inputs, targets in human_test_loader:
                        inputs = inputs.to(device)
                        targets = targets.to(device)

                        outputs = model(inputs)
                        loss = criterion(outputs, targets)
                        test_loss += loss.item()

                        _, predicted = torch.max(outputs.data, 1)

                        all_preds.extend(predicted.cpu().numpy())
                        all_targets.extend(targets.cpu().numpy())

                test_loss = test_loss / len(human_test_loader)

                unique_targets = np.unique(all_targets)
                accuracy = accuracy_score(all_targets, all_preds)

                print(f"Combined {model_type.upper()} Model on Human Data Loss: {test_loss:.4f}, Accuracy: {accuracy:.4f}")
                print("\nClassification Report:")
                print(classification_report(all_targets, all_preds, labels=unique_targets,
                                          target_names=['Empty', 'Stomping']))

                print(f"\n=== Evaluating Combined {model_type.upper()} Model on Car Test Data ===")
                # For car data, remap labels
                X_car_combined = X_car_test
                y_car_combined = y_car_test.clone()
                y_car_combined[y_car_combined == 1] = 2

                car_combined_dataset = GeophoneDataset3Class(X_car_combined, y_car_combined)
                car_combined_loader = DataLoader(car_combined_dataset, batch_size=32)

                # Special evaluation for 3-class model on car data
                model.eval()
                all_preds = []
                all_targets = []
                test_loss = 0

                with torch.no_grad():
                    for inputs, targets in car_combined_loader:
                        inputs = inputs.to(device)
                        targets = targets.to(device)

                        outputs = model(inputs)
                        loss = criterion(outputs, targets)
                        test_loss += loss.item()

                        _, predicted = torch.max(outputs.data, 1)

                        all_preds.extend(predicted.cpu().numpy())
                        all_targets.extend(targets.cpu().numpy())

                test_loss = test_loss / len(car_combined_loader)

                unique_targets = np.unique(all_targets)
                accuracy = accuracy_score(all_targets, all_preds)

                print(f"Combined {model_type.upper()} Model on Car Data Loss: {test_loss:.4f}, Accuracy: {accuracy:.4f}")
                print("\nClassification Report:")
                # Map labels for display
                target_names = []
                for label in unique_targets:
                    if label == 0:
                        target_names.append('Empty')
                    elif label == 2:
                        target_names.append('Car')
                print(classification_report(all_targets, all_preds, labels=unique_targets,
                                          target_names=target_names))


# Main execution
if __name__ == "__main__":
    print("="*100)
    print("GEOPHONE 3-CLASS MOVEMENT DETECTION ANALYSIS - REALISTIC VERSION")
    print("Models: LSTM, GRU, CNN | Preprocessing: Time-Domain, STFT")
    print("Scenarios: Human-Only, Car-Only, Combined")
    print("Evaluation: 80/20 train-test split + 5-fold cross-validation on full dataset")
    print("="*100)

    # Set paths
    if IN_COLAB:
        drive.mount('/content/drive')
        drive_path = '/content/drive/MyDrive/Geophone/'
    else:
        drive_path = './Geophone/'

    # Run all experiments
    all_results = run_all_experiments(drive_path)

    # Perform cross-domain evaluation
    perform_cross_domain_evaluation(drive_path, all_results)

    # Create summary visualization
    create_summary_table(all_results)

    # Plot model comparison
    if all_results:
        plot_model_comparison(all_results)

    # Save results
    results_path = os.path.join(drive_path, 'models', 'realistic_results_3class.json')
    results_to_save = {}
    for exp_name, results in all_results.items():
        if 'test_results' in results:
            results_to_save[exp_name] = {
                'test_accuracy': float(results['test_results']['accuracy']),
                'test_f1_mean': float(np.mean(results['test_results']['f1'])),
                'cv_accuracy_mean': float(results['cv_results']['accuracy_mean']),
                'cv_accuracy_std': float(results['cv_results']['accuracy_std']),
                'cv_f1_mean': float(results['cv_results']['f1_mean']),
                'cv_f1_std': float(results['cv_results']['f1_std'])
            }

    with open(results_path, 'w') as f:
        json.dump(results_to_save, f, indent=4)

    print(f"\nResults saved to: {results_path}")
    print("\nEvaluation Summary:")
    print("- Test Set Results: Performance on held-out 20% test set")
    print("- Cross-Validation Results: Average performance across 5 folds using entire dataset")
    print("\nAnalysis complete!")
```
